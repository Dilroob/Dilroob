```python
sns.lineplot(x='total_bill',y='tip',data=df)  # sns.lineplot(x,y,dataset)
```

```python
sns.lineplot(x='total_bill',y='tip',data=df,marker=True,style='sex',linewidth=3)
plt.title('tips based on total bill')
plt.grid()
```

```python
sns.set_style('dark')   #types(dark,darkgrid,white,ticks)
sns.lineplot(x='total_bill',y='tip',data=df,marker=True,style='sex',linewidth=3)
plt.title('tips based on total bill')
plt.grid()
```

```python
sns.set_style('dark')
sns.set_context('poster')
sns.lineplot(x='total_bill',y='tip',data=df,marker=True,style='sex',linewidth=3)
plt.title('tips based on total bill')
plt.grid()
```

```python
sns.set_style('dark')
sns.set_context('talk')
sns.lineplot(x='total_bill',y='tip',data=df,marker=True,style='sex',linewidth=3)
plt.title('tips based on total bill')
plt.grid()
```

```python
# boxplot

sns.boxplot(x='total_bill',data=df)
```

```python
sns.pairplot(df)     # to find the correalation btw numerical columns
```

```python
plt.figure(figsize=(15,10))

plt.subplot(2,2,1)
plt.title('tips based on total bill')
plt.grid()
sns.set_style('dark')
sns.set_context('talk')
sns.lineplot(x='tip',
             y='total_bill',
             data=df,
             hue='sex',
             palette={'Male':'purple','Female':'magenta'},
             marker=True,
             linewidth=3)

plt.subplot(2,2,2)
plt.title('tips based on total bill')
plt.grid()
sns.set_style('dark')
sns.set_context('talk')
sns.scatterplot(x='tip',
                y='total_bill',
                data=df,
                hue='sex',
                palette={'Male':'purple','Female':'magenta'},
                size='size',
                alpha=0.8)

plt.subplot(2,2,3)
plt.title('tips based on total bill')
plt.grid()
sns.set_style('dark')
sns.set_context('talk')
sns.boxplot(x='total_bill',
            data=df,
            hue='sex',
            palette={'Male':'purple','Female':'magenta'})

plt.tight_layout()
```

```python
# replot    used to show the rewlationship between variables
sns.relplot(x='size',
           y='total_bill',
           col='day',
           col_wrap=2,
           data=df,
           height=4,
           aspect=2)    # aspect  ------> width=height*1.5
```
