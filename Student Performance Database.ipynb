```python
import pandas as pd
```

```python
a={'Student_ID':[101,102,103,104,105,106,107,108],
'Name':['Ali','Sara','John','Emma','David','Liya','Arjun','Meera'],
'Age':[21,22,20,23,21,22,23,20],
'Department':['CS','IT','CS','EC','IT','CS','EC','IT'],
'Marks':[85,40,90,60,75,88,55,95],
'Attendance':[92,75,88,95,80,91,70,98]}
```

```python
df=pd.DataFrame(a)
df
```

## 1

```python
df[df['Marks']>85]
```

## 2

```python
df[(df['Attendance']>80) & (df['Department']=='IT')]
```

## 3

```python
df.sort_values(by='Marks',ascending=False)
```

## 4

```python
df[(df['Marks']>=60) & (df['Marks']<=90)]
```

## 5

```python
df[df['Age']==21][['Name','Department']]
```

## 6

```python
avg_marks=df['Marks'].mean()
df[(df['Department']=='CS') & (df['Marks']>avg_marks)]
```

## 7

```python
df.sort_values(by=['Department','Marks'],ascending=[True,False])
```

## 8

```python
min_attendance=df['Attendance'].min()
df[df['Attendance']==min_attendance]
```

## 9

```python
df[(df['Marks']>90) | (df['Attendance']>95)]
```

## 10

```python
sorted_marks=df['Marks'].sort_values(ascending=False)
second_highest=sorted_marks.iloc[1]
df[df['Marks']==second_highest]
```
