```python
def add_student_record():
    name = input("Enter student name: ")
    roll_no = input("Enter roll number: ")
    course = input("Enter course: ")
    with open("student_records.txt", "a") as file:
        file.write(f"{name},{roll_no},{course}\n")
    print("Student record added successfully")

def display_all_records():
    try:
        with open("student_records.txt", "r") as file:
            records = file.readlines()
            if not records:
                print("No records found")
            else:
                print("Student Records:")
                for record in records:
                    record = record.replace("\n", "")
                    parts = record.split(",")
                    if len(parts) == 3:
                        name, roll_no, course = parts
                        print(f"Name: {name}, Roll No: {roll_no}, Course: {course}")
    except FileNotFoundError:
        print("No records found")

def search_student_record():
    roll_no = input("Enter roll number to search: ")
    try:
        with open("student_records.txt", "r") as file:
            records = file.readlines()
            found = False
            for record in records:
                record = record.replace("\n", "")
                parts = record.split(",")
                if len(parts) == 3:
                    name, r_no, course = parts
                    if r_no == roll_no:
                        print(f"Name: {name}, Roll No: {r_no}, Course: {course}")
                        found = True
                        break
            if not found:
                print("Student record not found")
    except FileNotFoundError:
        print("No records found")

def count_total_records():
    try:
        with open("student_records.txt", "r") as file:
            records = file.readlines()
            print(f"Total student records: {len(records)}")
    except FileNotFoundError:
        print("No records found!")

def main():
    while True:
        print("\nStudent Record Management System")
        print("1. Add new student record")
        print("2. Display all student records")
        print("3. Search for a student")
        print("4. Count total student records")
        print("5. Exit")
        choice = input("Enter your choice: ")
        if choice == "1":
            add_student_record()
        elif choice == "2":
            display_all_records()
        elif choice == "3":
            search_student_record()
        elif choice == "4":
            count_total_records()
        elif choice == "5":
            break
        else:
            print("Invalid choice! Please try again.")

if __name__ == "__main__":
    main()
```
