```python
plt.figure(figsize=(20,20))
font={'family':'serif','color':'red','size':20}
c=['orange','gold','red','purple']
plt.suptitle('Car Sales over the years(1984-2015)',size=50,color='blue')

plt.subplot(3,2,1)
plt.plot(df['year'].head(20),df['sellingprice'].head(20),marker='*',color='blue',ms=18,mec='yellow',mfc='red',linewidth=5,linestyle='-.',alpha=0.8)
plt.title('Line Graph',color='green',size=30,fontdict=font)
plt.xlabel('names',fontdict=font)
plt.ylabel('marks',fontdict=font)
plt.grid(color='gold',linestyle='dotted')

plt.subplot(3,2,2)
plt.scatter(df['year'].head(20),df['sellingprice'].head(20),marker='*',s=200,color='orange',alpha=0.8)
plt.title('Scatter Graph',color='green',size=30,fontdict=font)
plt.xlabel('names',fontdict=font)
plt.ylabel('marks',fontdict=font)
plt.grid(color='gold',linestyle='dotted')

plt.subplot(3,2,3)
plt.bar(df['year'].head(20),df['sellingprice'].head(20),color=c,alpha=0.8,width=0.5)
plt.title('Bar Graph',color='green',size=30,fontdict=font)
plt.xlabel('names',fontdict=font)
plt.ylabel('marks',fontdict=font)

plt.subplot(3,2,4)
plt.barh(df['year'].head(20),df['sellingprice'].head(20),color=c,alpha=0.8,height=0.5)
plt.title('Hbar Graph',color='green',size=30,fontdict=font)
plt.xlabel('names',fontdict=font)
plt.ylabel('marks',fontdict=font)

plt.subplot(3,2,5)
plt.hist(df['sellingprice'].head(5),bins=2,color='black',edgecolor='red')
plt.title('histoGram',color='green',size=30,fontdict=font)

plt.subplot(3,2,6)
exp=[0,0.3,0,0,0]
plt.pie(df['year'].head(5),autopct='%1.0f%%',startangle=90,labels=df['year'].head(5),explode=exp,colors=c,wedgeprops={'edgecolor':'black'},textprops={'fontsize':15},shadow=True)
plt.title('Pie Chart',color='green',size=30,fontdict=font)
plt.legend(df['year'].head(5),loc='upper left',title='chart',bbox_to_anchor=[1,1])

plt.tight_layout()
plt.show()
```
